# NumPy and Matplotlib primer
Run each cell, change the numbers, and run it again. Exercises are at the end.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Creating arrays

```python
a = np.array([1, 2, 3, 4])
M = np.array([[1, 2, 3], [4, 5, 6]])
print(a, a.shape, a.dtype)
print(M, M.shape)
print(np.zeros(3), np.ones((2, 2)))
print(np.arange(0, 10, 2))      # start, stop (excluded), step
print(np.linspace(0, 1, 5))     # 5 evenly spaced points, both ends included
```

## 2. Indexing and slicing

```python
print(a[0], a[-1], a[1:3])
print(M[1, 2])       # row 1, column 2
print(M[:, 0])       # first column
print(a[a > 2])      # boolean mask
```

## 3. Vectorized maths
Operations apply to every element, so you rarely need loops.

```python
print(a + 10, a * 2, a ** 2)
print(a.sum(), a.mean(), a.std())
print(M.sum(axis=0))   # down the rows: one value per column
print(M.sum(axis=1))   # across the columns: one value per row
```

## 4. Broadcasting and matrix products
`*` is element-wise. `@` is matrix multiplication.

```python
print(M - M.mean(axis=0))          # subtract each column's mean
print(M.T)                         # transpose
print(M @ np.array([1, 0, -1]))    # matrix-vector product
print(M @ M.T)
```

## 5. Solving linear systems and random numbers

```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
v = np.array([3.0, 5.0])
sol = np.linalg.solve(A, v)
print(sol, A @ sol)

rng = np.random.default_rng(0)     # the seed makes results repeatable
print(rng.uniform(0, 1, 3))
print(rng.normal(0, 1, 3))
```

## 6. Line plots

```python
x = np.linspace(-3, 3, 200)
fig, ax = plt.subplots()
ax.plot(x, x**2, label="x squared")
ax.plot(x, np.abs(x), "--", label="|x|")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Two convex functions")
ax.legend()
plt.show()
```

## 7. Scatter plots, histograms and subplots

```python
xs = rng.uniform(0, 10, 100)
ys = 2 * xs + rng.normal(0, 2, 100)
fig, (left, right) = plt.subplots(1, 2, figsize=(10, 4))
left.scatter(xs, ys, s=15)
left.set_title("Scatter")
right.hist(ys - 2 * xs, bins=15)
right.set_title("Histogram of the noise")
plt.tight_layout()
plt.show()
```
